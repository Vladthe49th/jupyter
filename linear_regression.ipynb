# Навчаємо регресію продавати парасольки

```python


import numpy as np
from sklearn.linear_model import LinearRegression


X = np.array([[25], [20], [15]])
y = np.array([5, 10, 15])


model = LinearRegression()
model.fit(X, y)


temperature = np.array([[18]])
prediction = model.predict(temperature)


print("Температура:", temperature[0][0], "C")
print("Прогноз продажів:", prediction[0], "парасольок")
```
