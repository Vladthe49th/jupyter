```python
import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, gutenberg
from collections import Counter

# Download required NLTK resources
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('gutenberg')

print("Libraries imported successfully!")
```

## Part 1. Personal Sentence Analysis (About my day)

```python
# Our sentence
sentence = "Today I woke up early, survived my marazmatic math teacher at university, eaten some cheeze afterwards"

# Tokenization
tokens = word_tokenize(sentence)

# Convert tokens to lowercase
tokens = [word.lower() for word in tokens]

# Get English stop words
stop_words = set(stopwords.words('english'))

# Remove stop words and punctuation
clean_tokens = [
    word for word in tokens
    if word.isalpha() and word not in stop_words
]

# Display results
print("Original sentence:", sentence)
print("Tokens:", tokens)
print("Cleaned tokens:", clean_tokens)
print("Remaining tokens:", len(clean_tokens))
```

## Part 2. Alice in Wonderland  Analysis


```python
# Load text
alice_words = list(gutenberg.words('carroll-alice.txt'))

# Count all tokens
total_tokens = len(alice_words)

# Count alphabetic words
total_words = sum(word.isalpha() for word in alice_words)

print("Total tokens:", total_tokens)
print("Total alphabetic words:", total_words)
```

```python
# Convert words to lowercase
alice_cleaned = [
    word.lower()
    for word in alice_words
    if word.isalpha() and word.lower() not in stop_words
]

print("Words after cleaning:", len(alice_cleaned))
```

```python
# Count frequencies
word_frequencies = Counter(alice_cleaned)

# Get five most common words
most_common_words = word_frequencies.most_common(5)

print("Top 5 most frequent words:")

for word, frequency in most_common_words:
    print(f"{word}: {frequency}")
```
