# Нейронка по розпізнаванню котів та собак

```python
import os
from icrawler.builtin import BingImageCrawler
```

```python
# Создаємо датасетні директорії
os.makedirs("dataset/cats", exist_ok=True)
os.makedirs("dataset/dogs", exist_ok=True)

print("Датасетні директорії успішно створено!")
```

```python
# Завантажуємо картинки з кішками
cat_crawler = BingImageCrawler(
    storage={"root_dir": "dataset/cats"}
)

cat_crawler.crawl(
    keyword=" cats different breeds",
    max_num=100,
    min_size=(200, 200)
)

print("Картинки з кішками завантажено!")
```

```python
# Завантажуємо картинки, але з собачками

dog_crawler = BingImageCrawler(
    storage={"root_dir": "dataset/dogs"}
)

dog_crawler.crawl(
    keyword=" dogs different breeds",
    max_num=100,
    min_size=(200, 200)
)

print("Картинки з собаками завантажено!")
```

```python
# Перевіряємо завантажені картинки

image_extensions = (".jpg", ".jpeg", ".png", ".webp")

cats = [
    f for f in os.listdir("dataset/cats")
    if f.lower().endswith(image_extensions)
]

dogs = [
    f for f in os.listdir("dataset/dogs")
    if f.lower().endswith(image_extensions)
]

print(f"Кішок: {len(cats)}")
print(f"Собак: {len(dogs)}")
print(f"Всього: {len(cats) + len(dogs)}")
```
