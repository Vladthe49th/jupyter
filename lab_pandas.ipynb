# Завдання 3 - об'єднуємо клієнтів та замовлення

```python
clients = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Ім'я": [
        "Олександр",
        "Марія",
        "Іван",
        "Бінгус",
        "Біллі"
    ],
    "Місто": [
        "Київ",
        "Львів",
        "Одеса",
        "Інтернет-простір",
        "Америка (так так, я не помилився)"
    ]
})

print(clients)
```

```python
customer_orders = pd.DataFrame({
    "ID клієнта": [1, 2, 1, 3, 4, 2, 5, 5],
    "Замовлення": [
        "Ноутбук",
        "Смартфон",
        "Навушники",
        "Клавіатура",
        "Вазелин",
        "Блок для криптомайнингу",
        "Сколопендра",
        "Завелин"
    ],
    "Вартість": [
        1200,
        500,
        800,
        700,
        2000,
        1500,
        900,
        600
    ]
})

print(customer_orders)
```

```python
merged_orders = pd.merge(
    clients,
    customer_orders,
    left_on="ID",
    right_on="ID клієнта"
)

print(merged_orders)
```

### Вартість замовлень

```python
result = merged_orders.groupby(
    ["ID", "Ім'я", "Місто"],
    as_index=False
)["Вартість"].sum()

print(result)
```

```python
print("Загальна вартість усіх замовлень:", result["Вартість"].sum())
```
