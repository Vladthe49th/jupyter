```python

import pandas as pd
import matplotlib.pyplot as plt


data = {
    "day": list(range(1, 29)),

    "customer_rating": [
        4.2, 4.5, 4.1, 4.7, 4.6, 4.3, 4.8,
        4.4, 4.6, 4.9, 4.2, 4.5, 4.7, 4.3,
        4.8, 4.6, 4.4, 4.9, 4.5, 4.7, 4.1,
        4.6, 4.8, 4.3, 4.7, 4.5, 4.9, 4.6
    ],

    "cups_sold": [
        85, 92, 78, 105, 98, 88, 115,
        90, 108, 120, 82, 96, 112, 87,
        118, 103, 91, 125, 99, 110, 80,
        106, 122, 89, 114, 97, 128, 109
    ],

    "coffee_type": [
        "Espresso", "Latte", "Cappuccino", "Americano",
        "Latte", "Espresso", "Cappuccino", "Americano",
        "Latte", "Cappuccino", "Espresso", "Americano",
        "Latte", "Cappuccino", "Espresso", "Americano",
        "Latte", "Cappuccino", "Espresso", "Latte",
        "Americano", "Cappuccino", "Latte", "Espresso",
        "Cappuccino", "Americano", "Latte", "Cappuccino"
    ],

    "temperature": [
        68, 65, 67, 70, 66, 69, 68,
        71, 65, 67, 69, 70, 66, 68,
        69, 71, 65, 67, 70, 66, 71,
        68, 65, 69, 67, 70, 66, 68
    ]
}


df = pd.DataFrame(data)

display(df.head())


print("Розмір датасету:", df.shape)
print("\nІнформація про колонки:")
df.info()
```

# Графік 1 - оцінки клієнтів

```python


plt.figure(figsize=(10, 5))

plt.plot(
    df["day"],
    df["customer_rating"],
    color="green",
    marker="o"
)

plt.title("Customer Ratings Over 28 Days")
plt.xlabel("Day")
plt.ylabel("Customer Rating")

plt.grid(True)
plt.show()
```

# Графік 2 - точкова діаграма зв'язку продажів і рейтингу

```python


plt.figure(figsize=(10, 5))

plt.scatter(
    df["cups_sold"],
    df["customer_rating"],
    alpha=0.5
)

plt.title("Coffee Sales vs Customer Ratings")
plt.xlabel("Cups Sold")
plt.ylabel("Customer Rating")

plt.show()
```

# 3 - Гістограмма кількості продажів

```python

# Гістограма розподілу кількості проданих чашок

plt.figure(figsize=(10, 5))

plt.hist(
    df["cups_sold"],
    bins=8,
    color="coral",
    edgecolor="black"
)

plt.title("Distribution of Coffee Sales")
plt.xlabel("Cups Sold")
plt.ylabel("Frequency")

plt.show()
```

# 4- Середні продажі

```python


average_sales = df.groupby("coffee_type")["cups_sold"].mean()

# Виводимо середні продажі для кожного напою
print(average_sales)
```

```python


plt.figure(figsize=(10, 5))

average_sales.plot(kind="bar", color="coral", edgecolor="black")

plt.title("Average Sales by Coffee Type")
plt.xlabel("Coffee Type")
plt.ylabel("Average Cups Sold")

plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```
