```python
import numpy as np
```

# Завдання 1 - 1D массив

```python
arr = np.random.randint(1, 101, 10)

print("Масив:", arr)
print("Перший елемент:", arr[0])
print("Центральний елемент:", arr[5])
print("Останній елемент:", arr[-1])
```

# Завдання 2 - масив від 10 до 50 та зріз

```python
arr = np.arange(10, 51)
print(arr[::-3])
```

# Завдання 3 - матриця 4х4 та підматриця

```python
matrix = np.arange(1, 17).reshape(4, 4)

print("Матриця:")
print(matrix)

print("\nПідматриця 2x2 у правому верхньому куті:")
print(matrix[:2, 2:])
```

# Завдання 4 - матриця 5х5

```python
matrix = np.random.randint(1, 101, (5, 5))

print("Матриця:")
print(matrix)

print("\nЦентральний елемент:", matrix[2, 2])
print("Третій стовпчик:", matrix[:, 2])
```

# Завдання 5 - позитивні числа через булеву маску

```python
arr = np.random.randint(-10, 11, 10)

print("Масив:", arr)
print("Додатні значення:", arr[arr > 0])
```

# Завдання 6 - кількість чисел, кратних 3

```python
arr = np.arange(1, 21)

print("Масив:", arr)

count = np.sum(arr % 3 == 0)

print("Кількість чисел, що діляться на 3:", count)
```

# Завдання 7 - фільтрація за двома умовами

```python
arr = np.random.randint(0, 101, 12)

print("Масив:", arr)
print("Елементи від 20 до 80:", arr[(arr > 20) & (arr < 80)])
```

# Завдання 8 - заміна парних чисел на 0

```python
arr = np.arange(1, 13).reshape(3, 4)

print("Початковий масив:")
print(arr)

arr[arr % 2 == 0] = 0

print("\nМасив після заміни парних чисел:")
print(arr)
```
