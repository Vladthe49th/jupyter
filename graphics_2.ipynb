# Завдання 1 - Моделювання зарплати співробітників

```python
import numpy as np
import matplotlib.pyplot as plt


n = 50

np.random.seed(42)
years = np.random.uniform(0, 30, n)

basic_pay = 15000
more_salary = 1200
shum = np.random.normal(0, 4000, n)

zarplata = basic_pay + more_salary * years + shum


plt.figure(figsize=(8, 5))
plt.scatter(years, zarplata, color="blue",
            label="Співробітники")


x = np.linspace(0, 30, 100)
y = basic_pay + more_salary * x
plt.plot(x, y, color="red",
         label="Залежність зарплати від стажу")

plt.title("Зарплата співробітників залежно від стажу")
plt.xlabel("Стаж роботи")
plt.ylabel("Зарплата")
plt.legend()
plt.grid(True)
plt.show()
```

# Завдання 2 - Залежність витрат на маркетинг від виручки


```python

revenue = np.linspace(0, 1000000, 100)


max_spend = 200000
k = 0.000005


money_spend = max_spend * (1 - np.exp(-k * revenue))

plt.figure(figsize=(8, 5))
plt.plot(revenue, money_spend, color="green",
         label="Витрати на маркетинг")

plt.title("Залежність витрат на маркетинг від виручки")
plt.xlabel("Виручка компанії")
plt.ylabel("Витрати на маркетинг")
plt.legend()
plt.grid(True)
plt.show()
```

# Завдання 3 - Моделювання рівня стресу працівника



```python

hours = np.arange(1, 13)


urgency = np.random.randint(1, 6, len(hours))


stress = hours * 4 + urgency * 8


stress = np.clip(stress, 0, 100)


plt.figure(figsize=(8, 5))
plt.plot(hours, stress, marker="o", color="purple",
         label="Рівень стресу")

plt.title("Залежність стресу від робочого навантаження")
plt.xlabel("Відпрацьовані години")
plt.ylabel("Рівень стресу")
plt.ylim(0, 100)
plt.legend()
plt.grid(True)
plt.show()
```

# Завдання 4 - Залежність кількості покупок від ціни

```python

price = np.linspace(10, 100, 50)


noise = np.random.normal(0, 3, len(price))


quantity = 1000 / price + noise


quantity = np.clip(quantity, 0, None)


plt.figure(figsize=(8, 5))
plt.scatter(price, quantity, color="orange",
            label="Кількість покупок")


x = np.linspace(10, 100, 100)
y = 1000 / x
plt.plot(x, y, color="red",
         label="Обернена пропорційність")

plt.title("Залежність кількості покупок від ціни")
plt.xlabel("Ціна товару")
plt.ylabel("Кількість куплених одиниць")
plt.legend()
plt.grid(True)
plt.show()
```

# Завдання 5 -  Моделювання часу в дорозі



```python

hours = np.arange(6, 23)


base_time = 25

weather = np.random.choice(
    ["Сонячно", "Дощ", "Сніг"],
    size=len(hours)
)

travel_time = []

for i in range(len(hours)):
    time = base_time

   
    if 7 <= hours[i] <= 9 or 16 <= hours[i] <= 19:
        time += 20
    elif 10 <= hours[i] <= 15:
        time += 5
    else:
        time += 0

   
    if weather[i] == "Дощ":
        time += 10
    elif weather[i] == "Сніг":
        time += 20
    else:
        time += 0

    travel_time.append(time)


plt.figure(figsize=(9, 5))
plt.plot(hours, travel_time, marker="o",
         color="blue", label="Час у дорозі")

plt.title("Залежність часу в дорозі від часу доби та погоди")
plt.xlabel("Година доби")
plt.ylabel("Час у дорозі (хвилини)")
plt.xticks(hours)
plt.legend()
plt.grid(True)
plt.show()


for i in range(len(hours)):
    print(
        f"{hours[i]}:00 | Погода: {weather[i]:8} | "
        f"Час у дорозі: {travel_time[i]} хв"
    )
```

# Завдання 6 - Залежність оцінки ресторану від очікування та обслуговування



```python

waiting_time = np.arange(5, 61, 5)

service_quality = np.random.uniform(3, 5, len(waiting_time))


rating = 5 * (1 / (1 + waiting_time / 30)) * (
    service_quality / 5
)


plt.figure(figsize=(9, 5))
plt.plot(waiting_time, rating, marker="o",
         color="green", label="Оцінка ресторану")

plt.title("Оцінка ресторану залежно від очікування та обслуговування")
plt.xlabel("Час очікування (хвилини)")
plt.ylabel("Оцінка ресторану")
plt.ylim(0, 5)
plt.legend()
plt.grid(True)
plt.show()

for i in range(len(waiting_time)):
    print(
        f"Очікування: {waiting_time[i]:2} хв | "
        f"Якість обслуговування: {service_quality[i]:.2f} | "
        f"Оцінка: {rating[i]:.2f}/5"
    )
```
