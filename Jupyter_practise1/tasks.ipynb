## Завдання 1 - Аналіз температури

Обчислимо середнє, максимальне та мінімальне значення температури,
а також різниці між максом та міном.

```python
temperatures = [2, 5, 4, 6, 7, 3, 1]

average_temperature = sum(temperatures) / len(temperatures)
maximum_temperature = max(temperatures)
minimum_temperature = min(temperatures)
temperature_difference = maximum_temperature - minimum_temperature
```

## Завдання 2 - Аналіз щоденних витрат

витрати, а також кількість днів коли ми втратили дуже багато грошей (1000).

## Завдання 3 - Аналіз голосів

Підрахуємо кількість голосів за кожен фрукт, визначимо лідера та сформуємо топ-3.

```python
from collections import Counter

votes = [
    "apple",
    "banana",
    "apple",
    "orange",
    "banana",
    "apple",
    "apple",
    "apple",
    "orange"
]

vote_counts = Counter(votes)

leader = vote_counts.most_common(1)[0]
top_3 = vote_counts.most_common(3)
```

## Завдання 4 - Аналіз результатів з предметів

Порівняємо середні, мінімальні і максимальні оцінки зі стандартного шкільного набору предметів

```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "Math": math,
    "Physics": physics,
    "English": english
}

averages = {
    subject: sum(scores) / len(scores)
    for subject, scores in subjects.items()
}

minimums = {
    subject: min(scores)
    for subject, scores in subjects.items()
}

maximums = {
    subject: max(scores)
    for subject, scores in subjects.items()
}

best_subject = max(averages, key=averages.get)
```

| Предмет | Середній бал | Мінімальний бал | Максимальний бал |
|---|---:|---:|---:|
| Math | 79.0 | 60 | 92 |
| Physics | 77.6 | 65 | 90 |
| English | 91.4 | 85 | 100 |

Best subject - English
